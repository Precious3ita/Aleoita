```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("anandshaw2001/top-spotify-songs-in-countries")

print("Path to dataset files:", path)
```

```python
import requests
from bs4 import BeautifulSoup

url = "https://www.kaggle.com/datasets/anandshaw2001/top-spotify-songs-in-countries"
response = requests.get(url)

print("Response status code:", response.status_code)
print("Response content:")
print(response.text)
```

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(response.content, "html.parser")
print (soup.prettify)
```

```python
header = soup.find("title")
print(header.prettify)
header = soup.find("script")
print(header.prettify)
```
